```python
# Filter for Asian-origin cars
asian_cars = df[df['origin'] == 'Asia']

# Find the maximum fuel efficiency
max_fuel_efficiency = asian_cars['fuel_efficiency_mpg'].max()
print(max_fuel_efficiency)
```

```python
# Step 1: Calculate initial median of horsepower
median_before = df['horsepower'].median()
print(f"Initial Median: {median_before}")

mode_hp = df['horsepower'].mode()[0]
print(f"Most Frequent Value (Mode): {mode_hp}")

df['horsepower_filled'] = df['horsepower'].fillna(mode_hp)

median_after = df['horsepower_filled'].median()
print(f"New Median: {median_after}")

if median_after > median_before:
    print("Result: Yes, it increased")
elif median_after < median_before:
    print("Result: Yes, it decreased")
else:
    print("Result: No")
```

```python
asia_cars = df[df['origin'] == 'Asia']

X_df = asia_cars[['vehicle_weight', 'model_year']].head(7)

X = X_df.values

XTX = X.T @ X

XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

sum_w = w.sum()
print(f"Weight vector w: {w}")
print(f"Sum of w: {sum_w:.3f}")
```
